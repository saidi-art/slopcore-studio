# ACE-Step cloud studio on the free Colab T4

Before running anything: **Runtime > Change runtime type > T4 GPU > Save**.
Then run the cells top to bottom. Cell 3 stays busy while the studio is up (that is normal); its output prints the public link to open.


## 1. Check the GPU


```
!nvidia-smi
```

## 2. Install ACE-Step (about 5 minutes, one time)


```
!curl -LsSf https://astral.sh/uv/install.sh | sh

!git clone https://github.com/saidi-art/ACE-Step-1.5.git

%cd ACE-Step-1.5

!~/.local/bin/uv sync
```

## 3. Launch the studio

First run downloads the model weights (~10 GB), which takes a while. When it finishes, copy the **.gradio.live** link from the output and open it in your browser. Keep this cell running while you generate.

The XL upgrade wants 12-20 GB of VRAM, so stay on the base model here.


```
!~/.local/bin/uv run acestep --share
```

## 4. Stop the studio (when done)

Runtime > Interrupt execution, or just disconnect the runtime. Nothing keeps running on its own; the free Colab session is not your Google budget.
